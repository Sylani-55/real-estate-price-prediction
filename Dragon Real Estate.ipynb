## Scikit-learn Design

Primarily, three types of objects
1. Estimators - It estimates some parameter based on a dataset. Eg. imputer
    It has a fit method and transform method.
    Fit method - Fits the dataset and calculates internal paramenters
3. Transformers - transform method takes input and returns output based on the learnings from fir(). It also has a convenience function called fir_transform() which fits and then transforms.
4. Predictors - LinearRegression model is an example of predictor fit(), and predict() are two common functions. It also gives score function which will evaluate the predictions.

## Feature Scaling


Two types of feature scaling
1. Min-Max scaling(Normalization)
value-min/(max-min)
sklearn provides a class called MinMaxScaler for this

2. Standardization 
 (value-mean)/std
   sklearn provides a class called standard scaler for this

## Creatin a Pipeline

```python
from sklearn.pipeline import Pipeline
from sklearn.preprocessing import StandardScaler
my_pipeline = Pipeline([
    ('imputer', SimpleImputer(strategy="median")),
    ('std_scaler', StandardScaler()),
])
```

```python
housing_num_tr = my_pipeline.fit_transform(housing)
housing_num_tr
```

```python
print(housing_tr.shape)
```

```python
print(housing_tr.columns)
```

```python
print(housing_num_tr.shape)
```

## Selecting a desired model for Dragon Real Estates

here we will be training our model

```python
from sklearn.linear_model import LinearRegression
from sklearn.tree import DecisionTreeRegressor
from sklearn.ensemble import RandomForestRegressor
model = RandomForestRegressor()
#model = LinearRegression()
#model = DecisionTreeRegressor()
model.fit(housing_num_tr, housing_labels)
```

## Evaluating the model

```python
import numpy as np
from sklearn.metrics import mean_squared_error
housing_predictions = model.predict(housing_num_tr)
mse = mean_squared_error(housing_labels, housing_predictions)
rmse = np.sqrt(mse)
```

```python
rmse
```

## Using better evaluation technique - Cross Validation 

```python
from sklearn.model_selection import cross_val_score
scores = cross_val_score(model, housing_num_tr, housing_labels, scoring="neg_mean_squared_error", cv=10)
rmse_scores = np.sqrt(-scores)
```

```python
rmse_scores
```

```python
def print_scores(scores):
    print("Scores:", scores)
    print("Mean:", scores.mean())
    print("Standard deviation:", scores.std())
```

```python
print_scores(rmse_scores)
```

## Testing the model on test data

```python
X_test = strat_test_set.drop("medv", axis=1)
Y_test = strat_test_set["medv"].copy()
X_test_prepeared = my_pipeline.transform(X_test)
final_predictions = model.predict(X_test_prepeared)
final_mse = mean_squared_error(Y_test, final_predictions)
final_rmse = np.sqrt(final_mse)
print(final_predictions, list(Y_test))
```
